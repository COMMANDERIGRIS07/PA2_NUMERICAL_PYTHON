## PA2 - EXPERIMENT 2: NUMERICAL PYTHON (NUMPY)

```python
import numpy as np
```

## PROBLEM A

```python
# Creates a reproducible random 5×5 integer ndarray
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))

# Computation of the mean and standard deviation parameters
x_bar = X.mean()
sigma = X.std()

# Normalize array
X_normalized = (X - x_bar) / sigma

# Array Outputs
print("Array X:\n", X)
print("\nArray X_normalized:\n", X_normalized)
print("\nMean of X_normalized:", X_normalized.mean())
print("Standard Deviation of X_normalized:", X_normalized.std())

# Save result
np.save("X_normalized.npy", X_normalized)
```

## PROBLEM B

```python
# Creates the first 100 positive integers, cube every element, and reshaping the result into a 10 × 10 ndarray named C
C = (np.arange(1, 101) ** 3).reshape(10, 10)

# Filter the values that are divisible by 4 
div_by_4 = C[C % 4 == 0]

# Array Outputs
print("Shape of C:", C.shape)
print("\nArray div_by_4:\n", div_by_4)
print("\nNumber of selected elements:", div_by_4.size)

# Save result to file
np.save("div_by_4.npy", div_by_4)
```

## PROBLEM C

```python
# Creates a 6×6 ndarray named S containing the squares of the first 36 positive integers in increasing row-major order
S = (np.arange(1, 37) ** 2).reshape(6, 6)

# Calculates the mean of array S
S_mean = S.mean()

# Filter elements strictly greater than the mean
above_mean = S[S > S_mean]

# Output Arrays
print("Array S:\n", S)
print("\nMean of S (S_mean):", S_mean)
print("\nArray above_mean:\n", above_mean)
print("\nNumber of selected elements:", above_mean.size)

# Save result to file
np.save("above_mean.npy", above_mean)
```
